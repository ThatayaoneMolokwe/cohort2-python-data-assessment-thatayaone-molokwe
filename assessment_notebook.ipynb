```python
#SECTION D:STUDENT PERFORMANCE DATA ANALYSIS PROJECT

#TASK 1:NOTEBOOK SETUP AND DATA LOADING
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('StudentsPerformance.csv')
df.head()
```

```python
#TASK 2:DATA INSPECTION AND CLEANING
print("Shape:", df.shape)
print("First 10 columns:")
print(df.columns[:10].tolist())

print("Dataset information:")
df.info()
#block below checks for any missing values and duplicates
missing = df.isna().sum().sort_values(ascending=False)
missing_percent = (df.isna().mean() * 100).sort_values(ascending=False)

missing_report = pd.DataFrame({
    "missing_count": missing,
    "missing_percent": missing_percent.round(2)
})

missing_report[missing_report["missing_count"] > 0].head(5)
df.duplicated().sum()
```

```python
#TASK 3: FEATURE ENGINEERING
df["total_score"]=df["math score"]+df["reading score"]+df["writing score"]
df["average_score"]=(df["math score"]+df["reading score"]+df["writing score"])/3
df.head()

```

```python
#TASK 4: GROUPBY AND SUMMARY TABLES

gender_counts = df["gender"].value_counts()

gender_counts
```

```python
df.groupby("gender")[["math score", "reading score", "writing score"]].mean()
```

```python
df.groupby("race/ethnicity")[["math score", "reading score", "writing score"]].mean()
```

```python
df.groupby("parental level of education")[["math score", "reading score", "writing score"]].mean()
```

```python
df.groupby("lunch")[["math score", "reading score", "writing score"]].mean()
```

```python
#TASK 5: DATA VISUALIZATION
import matplotlib.pyplot as plt
#histogram
total_score = [218, 247, 278, 148, 229]

plt.figure(figsize=(8, 5))
plt.hist(total_score, bins=8, edgecolor="black", color="skyblue")

plt.title("SCORES IN MATH")
plt.xlabel("Scores")
plt.ylabel("frequency")
plt.grid(axis="y", linestyle="--", alpha=0.7)  # Optional gridlines for readability
plt.show()
```

```python
#BAR CHART SHOWING TOTAL SCORES FOR FEMALES AND MALES
plt.figure(figsize=(8, 5))
plt.bar(df["gender"], df["total_score"])
plt.title("Student Scores")
plt.xlabel("Student")
plt.ylabel("Score")
plt.xticks(rotation=45)
plt.show()
```

```python
#box plot
import matplotlib.pyplot as plt
math_score = [72, 69, 90, 47, 76]
reading_score = [72, 90, 95, 57, 78]
performance = [math_score, reading_score,]
plt.figure(figsize=(8, 5))
plt.boxplot(performance, tick_labels=['math score', 'reading score'], patch_artist=True)

# Add titles and labels
plt.title('Student Performance')
plt.xlabel('scores')
plt.ylabel('total Scores')

# Show grid and plot
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

```python
#heatmap showing performance of students in math,readin and writing categories
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

df = pd.read_csv("StudentsPerformance.csv")
numbers_df = df.select_dtypes(include=["number"])
corr_matrix = numbers_df.corr()
plt.figure(figsize=(8, 6))
sns.heatmap(corr_matrix, annot=True, cmap="coolwarm", fmt=".2f", linewidths=0.5)


plt.title("Student Performance")
plt.show()
```
